# python-01 Python 第一步：变量、类型与 print

## 像给计算结果贴标签
price = 20 把整数 20 绑定到变量 price。等号在这里是赋值；比较相等写 ==。print(price) 把内容显示出来，否则运行脚本可能没有可见输出。
## 四种先认识的类型
int 是整数，float 是小数，str 是文本，bool 是 True/False。'20' 是文本，20 是数字，不能随意相加。None 表示缺少一个值，与 0、空字符串不同。
## 读代码的方法
程序通常从上到下执行，后面的变量可用前面的结果。# 后面是注释。变量名用能解释业务的英文或拼音，避免单字母堆叠。先改数值观察输出，再尝试增加计算。

```python
price = 20
quantity = 3
revenue = price * quantity
print("收入：", revenue)
print("变量类型：", type(revenue))
```

# python-02 列表、字典、索引和切片

## 一串值和一组对应关系
列表 [10,20,30] 按顺序保存多个值，字典 {'organic':100,'paid':80} 按键查值。列表从 0 开始编号，a[0] 是第一个；a[-1] 是最后一个。切片 a[0:2] 取第 0、1 位，不包含第 2 位。
## 不要把容器混淆
元组 tuple 常表示固定组合，集合 set 存不同元素且不能靠它保留排序。len 求长度，sum 对数值求和。字典的键必须存在才能直接索引，也可用 .get(键,默认值)。
## 可变对象要留心
b=a 不会复制整个列表，两个名字指向同一对象；简单列表复制用 a.copy()。复杂嵌套复制另有细节，先能识别“修改共享对象”的问题。

```python
amounts = [100, 50, 80]
channels = {"organic": 100, "paid": 80}
print(amounts[0], amounts[-1], amounts[:2])
print(sum(amounts), len(amounts))
print(channels.get("new_channel", 0))
```

# python-03 条件、循环与缩进

## 如果满足条件，就执行
if 后面写条件和英文冒号，下面缩进四个空格。elif 是另一种情况，else 兜底。缩进决定哪些代码属于分支，不只是排版。
## 对每个元素重复同一动作
for amount in amounts 依次把列表元素交给 amount。range(3) 得到 0、1、2。while 在条件成立时持续运行，初学尽量先用有明确终点的 for，避免无限循环。
## 计数器怎么工作
先设 total=0，再每次 total += amount，相当于 total=total+amount。break 结束循环，continue 跳过本次剩余操作。能用向量化处理表时，Pandas 通常不必逐行循环。

```python
amounts = [100, 50, 0, 80]
total = 0
for amount in amounts:
    if amount > 0:
        total += amount
        print("计入：", amount)
print("合计：", total)
```

# python-06 筛选、排序与新列

## 布尔条件像一列开关
orders['status']=='paid' 得到每行 True/False，把它放进 orders[...] 就只保留 True 的行。多个条件用 &、|、~，每个比较都加括号，不用 Python 的 and/or 连接整列。
## loc 与 iloc
.loc[行条件,列名] 按标签选取；.iloc[0:3,0:2] 按位置选取，右端不包含。sort_values 指定排序列和升降序。
## 安全添加新列
筛选后准备修改，先 .copy()。新列可以用整列运算生成，不必逐行循环。避免 chained assignment：不要连续两次方括号后直接赋值，使用 .loc 或明确副本。

```python
paid = orders.loc[(orders["status"] == "paid") & (orders["amount"] >= 100)].copy()
paid["amount_after_10pct_discount"] = paid["amount"] * 0.9
print(paid.sort_values(["amount", "order_id"], ascending=[False, True]))
```

# python-07 groupby、聚合与 transform

## 从明细到用户摘要
groupby('user_id') 按用户分组，agg 可以同时求订单数、金额总和、均值。size 计行数，count 计非空值，nunique 计不同值；与 SQL 计数的细节对照理解。
## 想保留每行怎么办
transform('sum') 把每组汇总值广播回每行，相当于 SQL 的 SUM OVER(PARTITION BY)。agg 会缩成组级结果，transform 保留原来长度。
## 最新行与组内排名
先按用户、时间、订单号排序，再按 user_id 去重保留最后一条，可以取最新订单；并列时间由订单号打破。groupby()['amount'].rank(method='dense',ascending=False) 对应金额稠密排名；method='min' 类似 RANK，method='first' 按现有行顺序打破并列，因此使用前需排序。
## 两个默认值
groupby 常默认把空分组键排除，需要时用 dropna=False。as_index=False 让分组列保留为普通列，初学更容易继续连接。

```python
paid = orders.loc[orders["status"] == "paid"].copy()
summary = paid.groupby("user_id", as_index=False).agg(paid_orders=("order_id", "size"), gmv=("amount", "sum"))
paid["user_gmv"] = paid.groupby("user_id")["amount"].transform("sum")
paid["amount_rank"] = paid.groupby("user_id")["amount"].rank(method="dense", ascending=False)
latest = paid.sort_values(["user_id", "order_at", "order_id"]).drop_duplicates("user_id", keep="last")
print(summary)
print(paid[["order_id", "user_id", "amount", "user_gmv", "amount_rank"]].head())
print("每位用户最新订单：")
print(latest[["user_id", "order_id", "order_at"]])
```

# python-08 merge、concat 与连接检查

## 和 SQL JOIN 一样先检查键
merge(...,on='user_id',how='left') 左连接。validate='one_to_many' 表示期望左侧键唯一、右侧可重复；不满足时主动报错，有助于防止金额翻倍。
## 指标校验
连接前后比较订单数、总额和未匹配比例。indicator=True 增加匹配来源列。Pandas 缺失键之间的匹配行为与普通 SQL NULL 连接不同，遇到空键时要明确处理。
## 拼接与透视
concat 默认沿行方向拼接，列名不同会产生缺失。pivot 把唯一组合转为宽表，pivot_table 可先聚合；melt 把宽表拉成长表。行列结构是展示方式，不应偷偷改变指标定义。
## 实验按用户分析
先把 paid 订单汇成一人一行金额，再以 assignments 的入组用户为左表连接，没消费的金额填 0；is_buyer 在有 paid 订单时为 1（金额为零的支付订单也算买家）。最后按 arm 算用户数、购买率和人均金额，避免把订单当独立用户。

```python
paid = orders.loc[orders["status"] == "paid"]
joined = users.merge(paid, on="user_id", how="left", validate="one_to_many", indicator=True)
print(joined[["user_id", "order_id", "_merge"]])
print("合并后金额：", joined["amount"].sum())
print(paid.pivot_table(index="user_id", columns="category", values="amount", aggfunc="sum", fill_value=0))
spend = paid.groupby("user_id", as_index=False).agg(gmv=("amount", "sum"), paid_orders=("order_id", "size"))
experiment = assignments.merge(spend, on="user_id", how="left", validate="one_to_one")
experiment[["gmv", "paid_orders"]] = experiment[["gmv", "paid_orders"]].fillna(0)
experiment["is_buyer"] = (experiment["paid_orders"] > 0).astype(int)
print(experiment.groupby("arm").agg(users=("user_id", "size"), conversion=("is_buyer", "mean"), gmv_per_user=("gmv", "mean")))
```

# python-09 缺失、重复、日期与滚动

## 清洗要保留解释
isna() 找缺失，fillna() 补值，dropna() 删缺失。drop_duplicates(subset=[...]) 依据指定键去重，但先确认哪一行该保留，例如按更新时间排序保留最后一条。不能把真实重复购买删掉。
## 日期转成可计算的类型
pd.to_datetime 把字符串变日期时间；errors='coerce' 会把无效字符串变 NaT，随后必须检查数量。.dt.date 取日，.dt.to_period('M') 取月。统一时区后再比较。
## 排序后才移动
groupby().shift() 对应 LAG。rolling(7) 是七行；按时间索引的 rolling('7D') 是时间窗，端点由 closed 等参数控制。先补日历更容易解释每天的滚动金额。

```python
paid = orders.loc[orders["status"] == "paid"].copy()
paid["dt"] = pd.to_datetime(paid["order_at"]).dt.normalize()
daily = paid.groupby("dt")["amount"].sum()
days = pd.date_range("2026-01-01", "2026-01-10")
daily = daily.reindex(days, fill_value=0)
print(pd.DataFrame({"gmv": daily, "rolling7": daily.rolling(7, min_periods=1).sum()}))
print("城市缺失数：", users["city"].isna().sum())
```

# python-10 NumPy、向量化和可复现抽样

## 数组会整批运算
NumPy 的数组与 Python 列表不同，np.array([1,2])*2 得 [2,4]，列表 [1,2]*2 会重复列表。向量化用底层数组操作，通常比逐个 Python 循环更适合大表。
## 随机不是不可验证
np.random.default_rng(42) 创建固定种子的随机数生成器，同一算法环境下复现同一抽样流程。换种子可检验结论是否依赖某次抽样；种子不能消除抽样误差。
## 文件与图表在本讲义的安排
网页运行区只展示文本与表格，交互图由后面的专用控件提供。导出的 Notebook 可在 VS Code 中运行和画图；导出脚本会包含示例数据初始化，离开网页也能用。
## 条件分层和除零
np.where(条件,成立取值,否则取值) 可以整列打标签，多分支可用 np.select。计算 Pandas 比率前，用分母.where(分母.ne(0)) 把零变成缺失，再用分子.div(分母)，避免出现无穷大。不要把“不可计算”和“结果为零”混成一件事。

```python
import numpy as np
x = np.array([1, 2, 3])
print("数组乘法：", x * 2)
rng = np.random.default_rng(42)
print("抽 5 个 0 到 9 的整数：", rng.integers(0, 10, size=5))
print("均值：", x.mean())
print("分层：", np.where(x >= 2, "高", "低"))
numerator = pd.Series([10, 0])
denominator = pd.Series([100, 0])
print("安全比率：", numerator.div(denominator.where(denominator.ne(0))))
```
